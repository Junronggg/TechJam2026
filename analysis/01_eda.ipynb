# 视频时长 vs 用户行为

```python
log_early["duration_sec"] = log_early["duration_ms"] / 1000

log_early["duration_bucket"] = pd.cut(
    log_early["duration_sec"],
    bins=[0, 10, 20, 30, 60, 120, 300, float("inf")],
    labels=["0-10s", "10-20s", "20-30s", "30-60s",
            "1-2m", "2-5m", "5m+"]
)
```

```python
duration_behavior = (
    log_early
    .groupby("duration_bucket", observed=True)
    [["is_click", "long_view", "is_like"]]
    .mean()
)

duration_behavior
```

```python
duration_behavior.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.ylabel("Positive Rate")
plt.title("Behavior Rate by Video Duration")
plt.show()
```

# Watch ratio 和行为关系

```python
log_early["watch_ratio"] = (
    log_early["play_time_ms"]
    / log_early["duration_ms"].replace(0, pd.NA)
)

log_early["watch_ratio"] = log_early["watch_ratio"].clip(0, 2)
```

```python
import numpy as np

log_early["watch_ratio"] = (
    log_early["play_time_ms"]
    / log_early["duration_ms"].replace(0, np.nan)
)

log_early["watch_ratio"] = log_early["watch_ratio"].clip(0, 2)
```

```python
log_early["watch_ratio_bucket"] = pd.cut(
    log_early["watch_ratio"],
    bins=[0, 0.25, 0.5, 0.75, 1.0, 2.0],
    labels=["0-25%", "25-50%", "50-75%", "75-100%", "100%+"],
    include_lowest=True
)
```

```python
watch_behavior = (
    log_early
    .groupby("watch_ratio_bucket", observed=True)
    [["is_like", "is_follow", "is_comment", "is_forward"]]
    .mean()
)

watch_behavior
```

```python
watch_behavior.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.ylabel("Positive Rate")
plt.xlabel("Watch Ratio")
plt.title("Engagement Rate by Watch Ratio")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# User activity 和 engagement 的关系

```python
user_stats = (
    log_early
    .groupby("user_id")
    .agg(
        interactions=("video_id", "size"),
        click_rate=("is_click", "mean"),
        long_view_rate=("long_view", "mean"),
        like_rate=("is_like", "mean")
    )
)
```

```python
user_stats["activity_group"] = pd.qcut(
    user_stats["interactions"],
    q=4,
    labels=["Low", "Mid-Low", "Mid-High", "High"]
)
```

```python
user_stats.groupby("activity_group", observed=True)[
    ["click_rate", "long_view_rate", "like_rate"]
].mean()
```

# Item popularity 和 engagement rate 的关系

```python
video_stats = (
    log_early
    .groupby("video_id")
    .agg(
        interactions=("user_id", "size"),
        click_rate=("is_click", "mean"),
        long_view_rate=("long_view", "mean"),
        like_rate=("is_like", "mean")
    )
)
```

```python
video_stats["popularity_group"] = pd.qcut(
    video_stats["interactions"],
    q=4,
    labels=["Tail", "Mid-Low", "Mid-High", "Head"]
)
```

```python
video_stats.groupby(
    "popularity_group",
    observed=True
)[["click_rate", "long_view_rate", "like_rate"]].mean()
```

# Tag 和用户行为的关系

```python
video_basic = pd.read_csv(
    DATA_DIR / "video_features_basic_pure.csv"
)

merged = log_early.merge(
    video_basic[["video_id", "tag"]],
    on="video_id",
    how="left"
)
```

```python
tag_stats = (
    merged
    .groupby("tag")
    .agg(
        interactions=("user_id", "size"),
        click_rate=("is_click", "mean"),
        long_view_rate=("long_view", "mean"),
        like_rate=("is_like", "mean")
    )
)
```

```python
top_tags = (
    tag_stats
    .sort_values("interactions", ascending=False)
    .head(20)
)

top_tags
```

```python
top_tags["click_rate"].plot(
    kind="bar",
    figsize=(12, 5)
)

plt.ylabel("Click Rate")
plt.title("Click Rate of Top 20 Tags")
plt.show()
```
